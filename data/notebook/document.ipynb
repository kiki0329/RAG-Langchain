```python
##Simple text file
import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts={
     "../data/text_files/python_intro.txt":"""Introduction to Python

Python is a high-level, interpreted, and general-purpose programming language. It was created by Guido van Rossum and first released in 1991. Python is known for its simple and easy-to-understand syntax, which makes it suitable for beginners.

Python is a free and open-source language. It supports different programming styles, including object-oriented, procedural, and functional programming. Python is widely used in web development, data science, artificial intelligence, machine learning, automation, game development, and software development.

One of the main advantages of Python is that it requires less code compared to many other programming languages. It also has a large collection of built-in libraries and frameworks that make programming easier.

In simple words, Python is a powerful, flexible, and beginner-friendly programming language used to develop different types of applications."""
   

}

for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f :
        f.write(content)
print("Sample text files created!")
```

```python
### how to read this text
###TEXT LOADER
#from langchain.document_loaders import TextLoader

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")
document=loader.load()
print(document)
```

```python
###Directory Loader
from langchain_community.document_loaders import DirectoryLoader

#laod all the text files from the directory

Dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls= TextLoader,    # if it a pdf we can use PDf loader
    loader_kwargs={'encoding':'utf-8'},
    show_progress=False
)
documents=Dir_loader.load()
documents
```

```python
from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader
Dir_loader=DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf",
    loader_cls= PyMuPDFLoader,    # if it a pdf we can use PDf loader
    show_progress=False
)

pdf_documents=Dir_loader.load()
pdf_documents
```
